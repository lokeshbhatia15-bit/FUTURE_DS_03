```python
df['Channel'] = df['Channel'].fillna('Unknown')
df['Stage_Visited'] = df['Stage_Visited'].fillna('Not Specified')
```

```python
region_revenue = df.groupby('Region')['Revenue'].sum()

print(region_revenue)
```
